```python
# deep learning is an extremely modular system
# - able to produce an output from a gien input through a series of computations
# - with given desired output, 
#   able to pass on info to its inputs on how to change to arrive at the desired output (predominately gradient-based optimization)

# raw input data 
# => forward coputation:
#   -> representation learning (linear transformation + non-linear activation function)
#   -> linear classifier (linear transformation + output activation function)
#   -> computed output (loss function against desired output) 
# => backward computation  

# matrix multiplication can be thought of as a linear transformation of vectors
# stack mulitple linear and non-linear transformation on top of each other
# <=> essentially warping, twisting and squeezing the input vector space (with features) into another space
# the representation learning block learn the right transformations, which makes the task (seperating) easier 


# linear transformation: 
# - rotation, reflection, scaling and so on are purely linear beacuse the origin won't change (A*X^T)
# - in neural network, actually mean affine transformation, which moves vector space (A*X^T + b)

# activation function:
# - non-linear differentiable function, transform input vector space (linearly inseparable) to a linearly separable space
# - pupluar activation functions:
#   * sigmoid 
#      g(x) = 1/(1+e^{-x}, squashes the input between 0 and 1, 
#      the gradients tend to zero on the flat portions of sigmoid which stops learning and we call this staturating of the activation,
#      not typically used in deep learning except in the output layer
#   * hypoerbolic tangent (tanh)
#      tanh(x) = sinh(x)/cosh(x), outputs a value between -1 and 1,
#      similar to sigmoid, also a saturating function,
#      symetrical around the origin (0) works better (than sigmoid) with the opimization of neural network
#   * rectified linear units (ReLU)
#      g(x) = max(x, 0)
#      the way the activation sharply becomes zero (does not tend to zero) gives non-linearity (non-saturating)
#      advantages: cheap computation, fast training coverage, bring sparsity (turn off large majority of neutrons)
#      disadvantages: dead ReLUs(output<0 will not learn), poistivie bias
#   * leaky ReLU:
#      g(x) = x, if x>=0
#          = alpha*x, if x<0 (alpha is slope parameter, typically set to a very small value, ensure no dead ReLUs)
#   * parametrized ReLU: consider alpha as a hyperparameter (learnable parameter)

# output activation function:
# - sigmoid: work for binary classification
# - softmax: softmax(x) = e^{x_i}/sum(e^{x_i}), convert a vector of K real values into K positive real value (resembles prob)

# forward computation / propagation: 
#   porpagate the input all the way through the network to get the output
# backward computation / propagation: 
#   usethe loss function to assess how close or far we are from the desired output, used to calculate the gradient with respect to all the parameters 
```

```python
# implementation of gradient descent

# loss function: L(a, b) = (a-8)^2 + (b-2)^2

def gradient(a, b):
    return 2*(a-8), 2*(b-2)


# theta (vector of parameters) = theta - lr * gradient(a, b)

# maximum number of iterations
maximum_iterations = 100
# current iteration
current_iteration = 0
# learing rate
learning_rate = 0.1
# inital value of a, b
current_a_value = 28
current_b_value = 27
# now, all that is left is the actual process of gradient descent

while current_iteration < maximum_iterations:
    # mark down previous values
    previous_a_value = current_a_value
    previous_b_value = current_b_value
    # calculate the gradient at current values
    gradient_a, gradient_b = gradient(previous_a_value, previous_b_value)
    # adjust the parameters using the gradients
    current_a_value = previous_a_value - learning_rate * gradient_a
    current_b_value = previous_b_value - learning_rate * gradient_b
    # next iteration
    current_iteration = current_iteration + 1

current_a_value, current_b_value
# gradient descent finds a solution that is pretty accurate as a=8, b=2 would make the loss function zero

# 3 popular variants of gradient descent that are used in learning task: 
#   batch gradient descent, stochastic gradient descent(SGD), mini-batch gradient descent
```
